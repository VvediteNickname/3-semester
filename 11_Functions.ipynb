1. Напишите функцию, которая будет возвращать индекс минимального элемента в списке. Если их несколько, то вернет все минимальные положения.

```python
def min_index(l):
    '''
    AssertionError
        Входной параметр не является списком
        Входной параметр является пустым списком    
        Входной параметр содержит не только числовые данные
    '''
    assert isinstance(l, list), "Входной параметр не является списком"
    assert len(l) > 0, "Входной параметр является пустым списком"

    current_min = l[0]
    ind = 0
    for j in l: 

        assert isinstance(j, (int, float)) and not isinstance(j, bool), "Входной параметр содержит не только числовые данные"
        if j < current_min: 
            current_min = j

    indexes = [x for x, v in enumerate(l) if v == current_min]

    return indexes
```

```python
example_list = [5,-4,4556,123,44,-5,0,324,-5]
min_ind = min_index(example_list)
print(*min_ind)
```

2. Напишите функцию, которая возвращает отсортированный методом пузырька список. 

```python
def bubble_sort(l):
    '''
    AssertionError
        Входной параметр не является списком
        Входной параметр является пустым списком    
        Входной параметр содержит не только числовые данные
    '''
    assert isinstance(l, list), "Входной параметр не является списком"
    assert len(l) > 0, "Входной параметр является пустым списком"

    for _ in range(len(l)):
        for i in range(len(l)-1):
            assert isinstance(i, (int, float)) and not isinstance(i, bool), "Входной параметр содержит не только числовые данные"
            if l[i] > l[i+1]:
                l[i],l[i+1] = l[i+1],l[i]
    
    return l
```

```python
example_list = [68, 5, 76, 70, 21, 3, 94, 100, 65, 28]
#3 5 21 28 65 68 70 76 94 100
print(*bubble_sort(example_list))
```

```python
import random
def guess():
    victory = False
    a = random.randint(1,100)
    for x in range(10):
        b = int(input())
        if a == b:
            victory = True
            break
        elif a < b:
            print("Загаданное число меньше введённого.")
        elif a > b:
            print("Загаданное число больше введённого.")
    if victory == True:
        print(f"Загаданное число равно введённому. Вы победили за {x} попыток!")
    else:
        print("Вы проиграли. Загаданное число -", a)
```
